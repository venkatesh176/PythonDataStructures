```python
class Stack:
    def __init__(self):
        self.container = deque()
    def push(self,val):
        self.container.append(val)
    def pop(self):
        return self.container.pop()
    def peek(self):
        return self.container[-1]
    def is_empty(self):
        return len(self.container)==0
    def size(self):
        return len(self.container)
```

```python


def is_match(ch1,ch2):
    match_dict={
        ')': '(',
        ']': '[',
        '}': '{'
    }
    return match_dict[ch1]==ch2
```

```python
def is_balanced(s):
    st1 = Stack()
    for ch in s:
        if ch=='(' or ch=='{' or ch == '[':
            st1.push(ch)
        if ch==')' or ch=='}' or ch == ']':
            if st1.size() ==0:
              return False
            if not is_match(ch, st1.pop()):
                return False
    return st1.size() ==0
        
```

```python
if __name__ == '__main__':
    print(is_balanced("({a+b})"))
    print(is_balanced("))((a+b}{"))
    print(is_balanced("((a+b))"))
    print(is_balanced("((a+g))"))
    print(is_balanced("))"))
    print(is_balanced("[a+b]*(x+2y)*{gg+kk}"))
```

```python
import threading
import time

q = Queue()  
try:
  def place_order(orders):  
    for order in orders:
          print(f"place: {q.append(order)}")
          time.sleep(0.5)

  def serve_order():  
        while True:
         if q is not None:
          print(f"serve: {q.pop()}")
          time.sleep(1)

  if __name__ == '__main__':

   orders = ['pizza','samosa','pasta','biryani','burger'] 

   thread1 = threading.Thread(target=place_order, args=(orders,))
   thread2 = threading.Thread(target=serve_order,)

   thread1.start()
   thread2.start()

   thread1.join()
   thread2.join()

except IndexError as i:
   pass
      #  print(i)
  
print("finished")
```

```python
from collections import deque
class Queue:
    def __init__(self):
        self.ref = deque()
    def append(self,val):
        self.ref.appendleft(val)
        
    def pop(self):
        return self.ref.pop()
    def front(self):
        return self.ref[-1]
```

```python
q = Queue()
q.append("1")
```
